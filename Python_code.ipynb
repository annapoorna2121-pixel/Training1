```python
n=int(input("Enter total no. of monkeys:"))
k=int(input("Enter no. of banana monkey can eat:"))
j=int(input("Enter no. of peanut monkey can eat:"))
m=int(input("Enter total no. of banana:"))
p=int(input("Enter total no. of peanut:"))
if(k==0 or j==0):
  print("INVALID INPUT")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No. of monkeys in tree:",n)
```

```python
n=int(input("enter the max no of candies:"))
k=int(input("enter the min no of candies:"))
c=1
r=n
while(c==1):
  m=int(input("enter the amount of candies needed:"))
  if(m>=n):
    print("invalid input:")
  else:
    r=r-m
    print("the candies sold:",m)
    if(r<=k):
      r=n
    print("sold=",m)
    print("remaining=",r)
    c=int(input("do you want to continue(1 or 0):"))

```
