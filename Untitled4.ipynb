**reverse** **a** **number**

```python
n=int(input("Enter a number:"))
op=0
while(n>0):
  r=n%10
  op=(op*10)+r
  n=n//10
print(op)

```

**some** **of** **digit**

```python
n=int(input("enter a number:"))
s=0
while n>0:
  rem=n%10
  s=s+rem
  n=n//10
print(s)
```

**largest** **number**

```python
n=int(input("enter a number:"))
s=0
while n>0:
  a=n%10
  if a>s:
   s=a
  n=n//10
print(s)
```

**palindrome**

```python
n=int(input("Enter a number:"))
a=n
t=0
while(n>0):
  r=n%10
  t=(t*10)+r
  n=n//10
if(a==t):
  print("palindrome:")
else:
  print("not palindrome:")
```

**anagram**

```python
s1=input("enter the first string:")
s2=input("enter the second string:")
if sorted(s1)==sorted(s2):
  print("anagram:")
else:
  print("not anagram:")
```

**count** **character** **in** **a** **string**

```python
n=input("enter a string:")
ch=input("enter character to count:")
count=n.count(ch)
print(count)
```

**without**

```python
n=input("enter a string:")
ch=input("enter character to count:")
count=0
for i in n:
  if i==ch:
    count=count+1
print("count:",count)
```

**remove** **duplicate** **element** **in** **string**

```python
s=input("Enter the string: ")
result=""
for char in s:
  if char not in result:
    result+=char
print("After removing duplicates =",result)
```
