**smallest** **element**

```python
n=int(input("enter the number of elements"))
arr= []
for i in range (n):
    m=int(input("enter the element:"))
    arr.append(m)
smallest=min(arr)
print("smallest element:",smallest)
```

**largest** **element**

```python
n=int(input("enter the number of elements"))
arr= []
for i in range (n):
    m=int(input("enter the element:"))
    arr.append(m)
    print(arr)
print("largest element:",max(arr))
```

**reverse** **array**

```python
arr=[]
n=int(input("Enter the number of elements: "))
for i in range(n):
  a=int(input("Enter element: "))
  arr.append(a)
print('Reversed Array',arr[::-1])
```

**occurence**

```python
lst = []
x = int(input("Enter limit: "))
for i in range(x):
    lst.append(int(input("Enter item: ")))
y = int(input("Enter element: "))
count = lst.count(y)
print("Occurrences:", count)
```
